```python
#1 Sucesion Aritmética

a_1 = 200
d = 24
sucesion = []
for i in range(1):
    n = i + 36
    sucesion.append(a_1 + (n - 1) * d)
    print(f'El término {n} es {sucesion[i]}')
```

```python
#suma de los primeros 36 terminos de la sucesion
a_1 = 200
d = 24
sucesion = []
for i in range(36):
    n = i + 1
    sucesion.append(a_1 + (n - 1) * d)
    print(f'El término {n} es {sucesion[i]}')

suma = sum(sucesion)
print(f'La suma de los primeros términos de la sucesión es {suma}.')  
```

```python
#13
import numpy as np

A = np.fromfunction(lambda i, j: 2*(i+1) + 3*(j+1), (3,3))
B = np.fromfunction(lambda i, j: 10*(i+1) * (j+1), (3,3))

C = np.dot(A, B)
print(f'La matriz C es: \n\n{C}')
```

```python
#14
import numpy as np

A = np.array([[2, 1],
              [3, -2]])

A_inv = np.linalg.inv(A)

R = ([[10], 
      [8]])
solucion = np.dot(A_inv, R)
print(f'La solución del sistema es: \n\n{solucion}')
```
