```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return 0.7*x**2

def g(x):
    return 10*x

x=np.arange(0, 50, 0.01) 

plt.plot(x,  f(x), color= 'm', label = 'Atleta 1')
plt.plot(x, g(x), color= 'g', label = 'Atleta 2')
plt.ylim(0, 200)
plt.xlim(0, 20)
plt.title('Comparación de las distancias recorridas de dos atletas')
plt.xlabel('Tiempo(segundos)')
plt.ylabel('Distancia Recorrida (metros)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.optimize import fsolve

def f(x):
    return 0.7*x**2

def g(x):
    return 10*x

def inter(x):
    return f(x)-g(x)

x = np.linspace(0, 15, 15)
solucion = np.around(fsolve(inter, x),1)
interseccion = np.unique(solucion)
print(f'El(los) punto(s) de interseccion de las funciones está(n) en x = {interseccion}')
a=f(14.3)
b=g(14.3)
print(f"La distancia recorrida por el Atleta 1 a los 14,3 segundos es {round(a,0)}")
print(f"La distancia recorrida por el Atleta 2 a los 14,3 segundos es {round(b,0)}")
```
