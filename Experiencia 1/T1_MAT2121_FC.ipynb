2) en un mismo grafico trace las tres funciones considerando la pronosticacion de 14 horas de lluvia

```python
import numpy as np
import matplotlib.pyplot as plt

def V(t):
    return 0.9*t**2 + 12.5*t + 36

def S(t):
    return 0.95*t**2 + 16.8*t + 31

def C(t):
    return 1.1*t**2 + 19.4*t + 26

t = np.linspace(0, 14, 500)

plt.plot(t, V(t), label="Valparaíso",)
plt.plot(t, S(t), label="Santiago",)
plt.plot(t, C(t), label="Concepción",)

plt.title("Agua acumulada durante 14 horas de lluvia")
plt.xlabel("Horas transcurridas desde las 06:00")
plt.ylabel("Agua acumulada (mm)")
plt.xticks()
plt.grid()
plt.legend()
plt.show()


```

3) la pimera ciudad en recibir ayuda

```python
from scipy.optimize import fsolve

def valparaiso(t):
    return 0.9*t**2 + 12.5*t + 36

def santiago(t):
    return 0.95*t**2 + 16.8*t + 31

def concepcion(t):
    return 1.1*t**2 + 19.4*t + 26

def raiz_valparaiso(t):
    return valparaiso(t) - 100

def raiz_santiago(t):
    return santiago(t) - 100

def raiz_concepcion(t):
    return concepcion(t) - 100

t_valparaiso = fsolve(raiz_valparaiso, 4)[0]
t_santiago = fsolve(raiz_santiago, 4)[0]
t_concepcion = fsolve(raiz_concepcion, 4)[0]

tiempos = {"Valparaíso": t_valparaiso, "Santiago": t_santiago, "Concepción": t_concepcion}

print(f"Valparaíso: {t_valparaiso:.2f}" ,"horas")
print(f"Santiago: {t_santiago:.2f}", "horas")
print(f"Concepción: {t_concepcion:.2f}", "horas")

primera_ciudad = min(tiempos)
print("La primera ciudad que debe recibir ayuda es:", primera_ciudad)

```
