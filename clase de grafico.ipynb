```python
import numpy as np
import matplotlib.pyplot as plt
def f(x): 
    return (120*x + 2000)

x = np.arange(0, 150, 0.01)
plt.plot(x, f(x), label = 'f(x)=120*x + 2000')
plt.title('Costo en pesos del consumo de energia en kw/h')
plt.ylabel('Cobro del consumo de energia $')
plt.xlabel('energia consumida en kw/h')
plt.legend()
plt.grid()
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.show()
```

Calcular la imagen

```python
def f(x): 
    return (120*x + 2000)
x = 100
print(f"Para un consumo de {x} kw/h, el cobro es de {f(x)} pesos")
```

Calcular la imagen en listas


```python
def f(x): 
    return (120*x + 2000)
consumos = [100, 200, 300, 400, 500]
for x in consumos:
    print(f"Para un consumo de {x} kw/h, el cobro es de {f(x)} pesos")
```

calcular la imagen pero que el usuario ingrese el numero


```python
def f(x): 
    return (120*x + 2000)
consumo = float(input("ingrese el consumo de energia: "))
print(f"Para un consumo de {consumo} kw/h, el cobro es de {f(consumo)} pesos")
```

Calculando Pre imagenes

```python
import numpy as np
from scipy.optimize import fsolve

def f(x):
    return 120*x + 2000 - 68000

#Valor/es inicial/es de la aproximación
xo = np.linspace(0, 1000, 1) #ecuacion de primer grado, por lo que solo tiene una solucion
solucion = fsolve(f, xo)
print(solucion[0])
print(f'La preimagen de $68.000 es {solucion[0]} kw/h de consumo de energia')
```
