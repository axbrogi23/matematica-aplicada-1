# Enunciado

La empresa de Ciberseguridad Seguri-net está analizando dos métodos de encriptación para archivos desde $20$ MB hasta $100$ MB. En el primer método, por cada incremento de $2$ MB en el tamaño del archivo, el tiempo de encriptación se incrementa de manera constante en $1,34$ segundos, tal como muestra la tabla (Método A). Mientras que el segundo método (Método B) tiene un tiempo de encriptación, también en segundos, que varía de acuerdo a un modelo cuadrático.

#### Método A:
<center>

| Tamaño del archivo (MB) | Tiempo de encriptación (segundos) | 
|-------------------------|---------------------|
|$$21$$                  |$$2,07$$               |
|$$40$$                  |$$14,8$$              |
|$$66$$                  |$$32,22$$              |
|$$74$$                  |$$37,58$$              |
|$$93$$                  |$$50,31$$              |

</center>

#### Método B:


$$g(x)=0,01x^2 - 0,4x + 8$$

Con ayuda de Python resuelve las siguientes preguntas, considera dos decimales para responder:

1. ¿Qué tipo de función permite calcular el tiempo de encriptación si se usa el método A? Justifica tu respuesta. Además, determina la forma algebraica de la función que modela el tiempo de encriptación en función del tamaño del archivo del método A. 
1. Calcula el tiempo de encriptación para ambos métodos cuando el tamaño del archivo es $62$ MB. ¿Cuál método empleó menos tiempo?
3. ¿Para qué tamaño(s) de archivo el tiempo de encriptación es el mismo en ambos métodos? ¿Cuál es el tiempo de encriptación?
1. En términos de tiempo, ¿para qué tamaño de archivos es más conveniente el método B?

# Desarrollo

```python
#metodo A pregunta 1 codigo

import numpy as np

x = np.array([21, 40, 66, 74, 93])
y = np.array([2.07, 14.8, 32.22, 37.58, 50.31])

pendiente, intercepto = np.polyfit(x, y, 1)

print(f"La pendiente es {pendiente:.2f}")
print(f"El coeficiente de posición es {intercepto:.2f}")
print(f"La función es f(x) = {pendiente:.2f} x {intercepto:.2f}")
```

Respuesta pregunta 1, metodo a: 

El metodo A se modela con una función lineal porque el incremento del tiempo por cada aumento constante del tamaño del archivo es casi constante.
Ajustando los datos, la función es:

f(x) = 0,67x - 12,00

donde x está en MB y f(x) en segundos.

```python
#Metodo b pregunta 2 codigo

import numpy as np

def f(x):
    return 0.67*x - 12

def g(x):
    return 0.01*x**2 - 0.4*x + 8

tamaño = 62

tiempo_A = f(tamaño)
tiempo_B = g(tamaño)

print(f"Metodo A: {tiempo_A:.2f} segundos")
print(f"Metodo B: {tiempo_B:.2f} segundos")
print(f"El metodo mas rapido es el metodo a con {tiempo_A:.2f} segundos mientras que el metodo B tarda {tiempo_B:.2f} segundos")
```

Respuesta pregunta 2: 

Para 62 MB:
Método A: f(62) = 0,67(62) - 12,00 = 29,54 segundos
Método B: g(62) = 0,01(62)**2 - 0,4(62) + 8 = 21,64 segundos
El método B empleó menos tiempo.

```python
#pregunta 3 sacar el tamaño de archivo para que ambos métodos tarden lo mismo

from scipy.optimize import fsolve
import numpy as np

def igualdad(x):
    return (0.67*x - 12) - (0.01*x**2 - 0.4*x + 8)

xo = np.linspace(20, 100, 10)

solucion = np.round(fsolve(igualdad, xo), decimals=2)

interseccion = np.unique(solucion)

print(f"el tiempo de encriptacion es el mismo para los dos metodos: {interseccion}")
```

```python
#grafico pregunta 3

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 0.67*x - 12

def g(x):
    return 0.01*x**2 - 0.4*x + 8

x = np.arange(20, 100.1, 0.1)
plt.plot(x, f(x), label="Metodo A", color="blue")
plt.plot(x, g(x), label="Metodo B", color="red")

plt.title("Tiempo de encriptacion segun tamaño del archivo")
plt.ylabel("Tiempo de encriptacion (segundos)")
plt.xlabel("Tamaño del archivo (MB)")
plt.legend()
plt.grid()
plt.axhline(0, color="black")
plt.axvline(0, color="black")
plt.show()
```

Respuesta pregunta 3:
Al igualar ambos metodos, se obtiene:

0,67x - 12 = 0,01x**2 - 0,4x + 8

Resolviendo, los tamaños donde tardan igual son:

24,14 MB
82,87 MB
En esos puntos, el tiempo de encriptacion es:

4,17 segundos
43,52 segundos
Por lo tanto, ambos metodos tienen el mismo tiempo de encriptacion en 24,14 MB y 82,87 MB.

Respuesta pregunta 4:

El metodo B es mas conveniente para tamaños de archivo entre 24,14 MB y 82,87 MB, ya que en ese intervalo tarda menos que el metodo A.
