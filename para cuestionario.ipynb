```python
#Pregunta 1

import numpy as np
from scipy.optimize import fsolve

def s(x):
    return -1.6 * x**2 + 1.3 * x + 141.3 #velocidad en kmH moto

def r(x):
    return 1.4 * x**2 - 5.6*x + 12.6 #velocidad en kmH auto

def interseccion(x):
    return s(x) - r(x)

xo = np.linspace(-50, 100, 2)
solucion = fsolve(interseccion, xo)
print(f"La intersección de las funciones es para x = {solucion} km/h")

```

```python
#pregunta 3
import numpy as np
from scipy.optimize import fsolve

def f(x):
    return 50*np.exp(0.3*x) - 551

#Valor/es inicial/es de la aproximación
xo = np.linspace(0, 100, 1)
solucion = np.around(fsolve(f, xo), 1)
print(f"Para 551 terabytes, se almacenan en: {solucion} meses")
```

```python
#pregunta 8

import numpy as np
from scipy.optimize import fsolve

def C(n):
    return 50 * n + 200 - 1500

xo = np.linspace(0, 100, 1)
solucion = np.around(fsolve(C, xo), decimals=2)

print(f"Fueron procesados: {solucion} terabytes de datos")
```

```python
#pregunta 15
import numpy as np
from scipy.optimize import fsolve

def I(x):
    return 50 * x + 0.5 * x**2

def C(x):
    return 10 * x + 450

def intercepto(x):
    return I(x) - C(x)

# Valor inicial de la aproximacion
xo = np.linspace(0, 20, 1)
solucion = np.around(fsolve(intercepto, xo), decimals=2)

print(f"El punto de equilibrio es de: {solucion} miles de suscriptores")
```
