```python
# pregunta 5
from scipy.optimize import fsolve

def a(t):
    return 0.5*t**2+4*t+10 #servidor a

def b(t):
    return 0.5*t**2 + 6*t + 2 #servidor b

def intercepto(t):
    return a(t) - b(t)

solucion = np.around(fsolve(intercepto, 1), decimals=2)
print(F"el momento es :" , solucion[0])
```

```python
#Pregunta 6
from scipy.optimize import fsolve

def ingreso (x):
    return 80*x + 2.5 *x**2

def costo (x):
    return 40*x + 2100

def ecuacion(x):
    return costo(x) - 3700

suscripcion = fsolve(ecuacion, 10)[0]
ingreso_total = ingreso(suscripcion)
print("cantidad de suscriptores: ", suscripcion)
print("cantidad de ingreso: ", ingreso_total)
```
