<br>

Se sabe que por un error de calibración de los sensores, las medidas registradas deberían ser un $5\%$ menores que las registradas.

# Problema 8

Una empresa tecnológica vende software y hardware en tres productos: $P_{1}$, $P_{2}$ y $P_{3}$, en cuatro zonas de la capital: norte, sur, este y oeste. Las ventas del mes, en millones de pesos, están registradas en las matrices $S$ y $H$, para software y hardware respectivamente.

<br>

\begin{split}
S = \begin{pmatrix}
 260 &  480 &  180  &  90  \\
 320 &  440 &  250  &  280 \\
 240 &  360 &  380  &  250  \\
\end{pmatrix} \text{ } H = \begin{pmatrix}
 360 &  250 &  300  &  250  \\
 450 &  500 &  350  &  380 \\
 290 &  300 &  460  &  400  \\
\end{pmatrix}
\end{split}

<br>

1. Calcule en la matriz $T$, del total de ventas de la empresa tecnológica.
2. Un trabajador indica que este mes, la sucursal de la zona sur recaudó más por la venta hardware que software, ¿es correcta su afirmación?
3. Debido al black friday se espera que el siguiente mes, las ventas de software aumenten un $7\%$ y las de hardware aumenten un $3\%$ respecto al mes actual. Encuentra en total de ventas esperadas para el siguiente mes y almacénalo en la matriz $M$. Utiliza *Python* para realizar los cálculos.
4. Interpreta los elementos $t_{24}$ y $m_{31}$ en el contexto del problema.
5. Calcula la matriz $M^{T}$ e interpreta el elemento $m_{41}^{T}$ en el contexto del problema.


```python
#1 total de ventas con la Matriz T

import numpy as np

S = np.array([
    [260, 480, 180,  90],
    [320, 440, 250, 280],
    [240, 360, 380, 250]
])

H = np.array([
    [360, 250, 300, 250],
    [450, 500, 350, 380],
    [290, 300, 460, 400]
])

T = S + H
print(f"Matriz T, Total de ventas:")
print(T)
```

```python
#2
software_sur = S [:, 1].sum()
hardware_sur = H [:, 1].sum()

print("Software en zona sur:", software_sur)
print("Hardware en zona sur:", hardware_sur)

#Respuesta = La afirmacion del trabajador es incorrecta, la zona sur recaudó mas por software que en hardware
```

```python
#3 para el proximo mes en black friday las ventas de software se incrementan en un 7% y las de hardware en un 3% respecto al mes actual.

M = 1.07 * S + 1.03 * H
print(f"Matriz M, El total de ventas para el próximo mes:")
print(f"\n{M}")
```

Respuesta 4 (Problema 8)

- \(t_{24}=660\): este mes, las ventas totales del producto \(P_2\) en la zona oeste fueron de **660 millones de pesos**.
- \(m_{31}=555{,}5\): para el próximo mes, se esperan ventas totales del producto \(P_3\) en la zona norte de **555,5 millones de pesos**.

```python
#5 calcula la matriz Mt e interpeta el elemento mt41 en el contexto del problema
MT = M.T

print(f"Matriz transpuesta de M:")
print(f"\n{MT}")
print(f"\n")
print(f"Elemento mt41: {MT[3, 0]}")
```
