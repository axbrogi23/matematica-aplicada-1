```python
#1 determinar si los paises mejoraron su desempeño en cada disciplina durante la 2da semaana con respecto a la primera semana
import numpy as np

A = np.array([[9.1, 9.0, 9.2],
             [9.3, 9.4, 8.6],
             [9.0, 8.9, 9.3]])

B = np.array([[9.5, 9.6, 9.1],
             [9.8, 9.5, 9.2],
             [8.9, 9.0, 9.4]])



D = B - A

print("Matriz D:")
print(D)

print(f"Elemento d23: {D[1, 2]:.1f}")

print(f"\n")

# 2. Proyección de mejora del 4% respecto a la primera semana
E = np.round(A * 1.04, 2)

print("Matriz E:")
print(E)
print(f"Elemento e21: {E[1, 0]:.2f}")

```

```python
#3 Paises ganador de las olimpiadas

import numpy as np

F = np.array([[41, 39, 28],
              [37, 33, 21],
              [28, 19, 18]])

G = np.array([[3],
              [2],
              [1]])

H = np.dot(F, G)
print(f"Matriz H el puntaje total obtenido por cada pais: \n {H}")


diferencia = H[0][0] - H[1][0]
print(f"La diferencia h11 - h21 es: {diferencia}")
```

```python
#4 Cantidad de medallas obtenidas en oro, plata y bronce
import numpy as np

A = np.array([[1, 1, 1],
              [3, 2, 1],
              [54, 37, 21]])

R = np.array([[10],
              [18],
              [341]])

A_inv = np.linalg.inv(A)
solucion = np.dot(A_inv, R)

print(f"La cantidad de medallas de oro obtenidas es: {solucion[0,0]:.0f}")
print(f"La cantidad de medallas de plata obtenidas es: {solucion[1,0]:.0f}")
print(f"La cantidad de medallas de bronce obtenidas es: {solucion[2,0]:.0f}")
```
