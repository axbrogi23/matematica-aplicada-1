```python
# Pregunta 1: Calcular el termino de la posicion 11
n_1 = 11
a_11 = 48.5 + 0.5 * n_1
print(f'El termino de la posición 11 es: {a_11:.0f}')

# Pregunta 2: Determinar si un determinado valor es termino de la sucesión
final = 1704 
termino = 0
n_2 = 0
dif = final - termino

while dif > 0:
    n_2 += 1
    termino = (n_2**3) - (2*n_2) 
    dif = final - termino

if dif == 0:
    print(f'El {final} es un termino de la sucesion y ocupa el lugar {n_2}')
else:
    print(f'El {final} no pertenece a la sucesion')
```

```python

# Pregunta 6: Suma de términos de una sucesión geométrica
sucesion = []
for i in range(12): # 12 meses en total
    n = i + 1
    termino = 3 * (2**(n - 1)) 
    sucesion.append(termino)

suma = sum(sucesion)
print(f'La suma de los primeros 12 términos de la sucesión es {suma}.') 
```

```python
import numpy as np

A = np.array([[800, 1600],
              [1, 1]])

B = np.array([[80000],
              [80]])

A_inv = np.linalg.inv(A)

# Multiplicacion de la matriz inversa para asi obtener el valor de X
X = np.dot(A_inv, B)

print(f"La matriz de (x, y) es: \n{X}")
```
