```python
b=[6, 9, 12, 15, 18]
# Con esta instrucci ́on se recorre el largo de la lista.
# El largo de la lista se define con la funci ́on len(b).
for i in range (len(b)):
n=i+1
print(f"b_{n}={b[i]}")
```

```python
#ejemplo 1 de la guia 4 de sucesiones 
#Creación de sucesiones con listas
sucesion = []
for i in range(6):
    n = i + 1
    sucesion.append(3*n)
    print(f'El término {n} es {sucesion[i]}')
```

```python
#ejemplo 2 de la guia 4 de sucesiones 
sucesion = []
for i in range(6):
    n = i + 10
    sucesion.append(5*0.92**n)
    print(f'El termino {n} es {round(sucesion[i], 1)}')
```

```python
#Ejemplo 3 de la guia 4 de sucesiones

final  = 10807
termino = 0
n = 0
dif = final - termino

while dif > 0:
    n += 1
    termino = 3*n**2+7
    dif = final - termino

if dif==0:
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```
